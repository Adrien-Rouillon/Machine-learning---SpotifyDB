## 2 · The libraries we'll use
Python for data science leans on four libraries — all included with Anaconda.

- **pandas** — the library for **tabular data** (rows and columns). Its main object is the **DataFrame**, essentially a spreadsheet you drive with code: load a file, filter rows, clean columns, compute summaries. This is where most of your time goes.
- **NumPy** — fast **numerical arrays** and math; the engine under pandas and scikit-learn.
- **Matplotlib** — **charts and plots**, so you can *see* your data.
- **scikit-learn** — the **machine-learning** toolkit (models + tools to train and evaluate them). We start using it in Week 3.

Standard nicknames you'll see everywhere:
`import pandas as pd` · `import numpy as np` · `import matplotlib.pyplot as plt`
